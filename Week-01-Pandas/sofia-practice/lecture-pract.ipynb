## BASIC MATH

```python
# up the price by 100
df['new_price'] = df['price'] * 100 # cretaes a new col new_price and has prices * 100 in there
df.head()
```

```python

cols = ["new_price", "price"]
df[cols]
```

```python
df['price'].mean()
# average value of the prices
```

calculated col operation:

```python
df['min_booking_amount'] = df['price'] * df['minimum_nights']
# looks at price col and min nights for every row, multiplies them (vectorized math)
# makes a new col called min_booking amt
df.head(2) # print first 2 rows
```

```python
cols = ["min_booking_amount", 'price', 'minimum_nights']
df[cols]
```
